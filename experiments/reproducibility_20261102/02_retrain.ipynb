# Optional retraining

Default: no training or data writes. First read README.md. Retraining needs private trusted input CSVs/caches and the verified reconstructed split directory. Published hashes and aggregate receipts cannot reconstruct missing raw artifacts. Use fresh destinations. The existing runner checks isolation, hashes, caches, validation selection and completion receipts.

```python
from pathlib import Path
import os,sys,subprocess
RUN_TRAINING=False
ROOT=Path.cwd()
if not (ROOT/'experiments/cluster_disjoint/train.py').exists(): ROOT=ROOT.parents[1]
INPUTS=Path(os.environ.get('SNOOPPI_INPUTS','/replace/with/private/inputs'))
SPLITS=Path(os.environ.get('SNOOPPI_FROZEN_SPLITS','/replace/with/verified/splits'))
OUTPUT=Path(os.environ.get('SNOOPPI_NEW_RUN','/replace/with/new/output'))
if RUN_TRAINING:
 assert INPUTS.is_dir() and (SPLITS/'manifest.json').is_file()
 assert not OUTPUT.exists(), 'Choose a fresh output directory'
 for seed in [44,45,46]:
  subprocess.run([sys.executable,str(ROOT/'experiments/cluster_disjoint/train.py'),'--inputs',str(INPUTS),'--splits',str(SPLITS),'--out',str(OUTPUT/f'seed{seed}'),'--seed',str(seed)],cwd=ROOT,check=True)
else:
 print('Training disabled. Configure the private input/split paths and explicitly enable RUN_TRAINING to fit models.')
```
